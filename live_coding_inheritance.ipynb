# Inheritance in Object-Oriented Programming


**Inheritance** is one of the fundamental pillars of Object-Oriented Programming. It allows you to create hierarchical relationships between classes, where **child classes** (also known as **subclasses** or **derived classes**) inherit attributes and methods from their **parent classes** (also called **superclasses** or **base classes**).

## Why Use Inheritance?

1. **Code Reusability** - Write once, use many times! No need to duplicate code.
2. **Maintainability** - Fix bugs in one place, and all child classes benefit.
3. **Extensibility** - Easily add new features without modifying existing code.
4. **Natural Modeling** - Represents real-world "is-a" relationships.

---

## Real-World Analogy: From Panda to Kung Fu Panda🐼➡️🥋

Imagine you're designing a video game with different types of pandas:

**Base Panda (Parent Class):**
- Has basic characteristics: `name`, `energy`, `bamboo_eaten`
- Can perform basic actions: `eat()`, `nap()`
- This represents what **ALL** pandas have in common

**Kung Fu Panda (Child Class):**
- **Inherits everything** from Base Panda (still eats, naps, has energy)
- **Adds new features**: `skill_level`, `train()`, `fight()`
- **Customizes behavior**: meditation instead of regular napping (method overriding)

This is an **"is-a" relationship**: A Kung Fu Panda **is a** Panda, but with special abilities!

## Python Syntax for Inheritance

Creating a child class in Python is simple:

```python
# Parent class
class Panda:
    pass

# Child class inherits from Panda
class KungFuPanda(Panda):
    pass
```

**Key points:**
- Put the parent class name in parentheses: `class Child(Parent):`
- Use `super()` to access parent class methods
- Child automatically gets all parent attributes and methods

Let's see this in action! 


````python
# Our Lovely Panda
class Panda:
    """Base class for a Panda"""
    def __init__(self, name, energy=50):
        self.name = name
        self.energy = energy
        self.bamboo_eaten = 0
        print(f"🐼 {self.name} the Panda is created!")
        
    def __repr__(self):
        return f"Panda(name={self.name}, energy={self.energy}, bamboo_eaten={self.bamboo_eaten})"
    
    def eat(self, bamboo_amount=1):
        """Panda eats bamboo and gains energy"""
        self.bamboo_eaten = self.bamboo_eaten +  bamboo_amount
        self.energy = self.energy +  bamboo_amount * 10
        print(f"{self.name} ate {bamboo_amount} bamboo. Energy: {self.energy}")
        
    def nap(self, duration=1):
        """Panda takes a nap and gains energy"""
        self.energy = self.energy +  duration * 20
        print(f"{self.name} napped for {duration} hour(s). Energy: {self.energy}")
````

### Using `super()` to extend the parent class

The `super()` function allows the child class to call methods from the parent class. This is useful when you want to:
1. Keep the parent's behavior
2. Add additional functionality on top

Let's add a **skill_level** attribute to our Kung Fu Panda and use `super()` to initialize the base Panda attributes first.

````python
# Let's extend the KungFuPanda class with additional attributes
class KungFuPanda(Panda):
    def __init__(self, name, energy=50, skill_level = 1):
        super().__init__(name, energy)
        self.skill_level = skill_level
    
    def __repr__(self):
        return super().__repr__() + " " + f"skill-level = {self.skill_level}"
        


````

````python
kevin = KungFuPanda(name= "Kevin")
````

````python
kevin.skill_level, kevin.name, kevin.energy, kevin.skill_level
````

### Polymorphism: Method Overriding

**Polymorphism** means "many forms". In OOP, it allows child classes to have methods with the same name as the parent but with different behavior.

Let's override the `nap()` method - Kung Fu Pandas train during their "naps"!

````python

class KungFuPanda(Panda):
    def __init__(self, name, energy=50, skill_level = 1):
        super().__init__(name, energy)
        self.skill_level = skill_level
    
    def __repr__(self):
        return super().__repr__() + " " + f"skill-level = {self.skill_level}"
    
    def nap(self, duration):
        super().nap(duration)
        super().nap(duration)
````

````python
kevin =KungFuPanda(name="Kevin")
kevin
````

````python
kevin.nap(duration=3)
````
